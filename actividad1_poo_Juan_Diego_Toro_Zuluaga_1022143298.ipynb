### Ejercicio Resuelto N° 4 - Edades de la familia

A la mamá de Juan le preguntan su edad, y contesta: tengo 3 hijos, pregúntele a Juan su edad. Alberto tiene 2/3 de la edad de Juan, Ana tiene 4/3 de la edad de Juan y mi edad es la suma de las tres. Hacer un algoritmo que muestre la edad de los cuatro.

````python
class Familia:

    def __init__(self, edad_juan):
        self.edad_juan = edad_juan

    def edad_alberto(self):
        return (2 / 3) * self.edad_juan

    def edad_ana(self):
        return (4 / 3) * self.edad_juan

    def edad_mama(self):
        return self.edad_juan + self.edad_alberto() + self.edad_ana()


edad_juan = float(input("Ingrese la edad de Juan: "))

familia = Familia(edad_juan)

print("Edad de Juan:", familia.edad_juan)
print("Edad de Alberto:", familia.edad_alberto())
print("Edad de Ana:", familia.edad_ana())
print("Edad de la mamá:", familia.edad_mama())
````

### Ejercicio Resuelto N° 5 - Seguimiento de instrucciones

Hacer un seguimiento (prueba de escritorio) del siguiente grupo de instrucciones:

```
INICIO
    SUMA = 0
    X = 20
    SUMA = SUMA + X
    Y = 40
    X = X + Y ** 2
    SUMA = SUMA + X / Y
    ESCRIBA: "EL VALOR DE LA SUMA ES:", SUMA
FIN_INICIO
```

````python
class PruebaEscritorio:

    def __init__(self):
        self.suma = 0
        self.x = 0
        self.y = 0

    def ejecutar(self):
        self.suma = 0
        self.x = 20
        self.suma = self.suma + self.x
        self.y = 40
        self.x = self.x + self.y ** 2
        self.suma = self.suma + self.x / self.y
        return self.suma


seguimiento = PruebaEscritorio()
resultado = seguimiento.ejecutar()

print("EL VALOR DE LA SUMA ES:", resultado)
````

### Ejercicio Propuesto N° 12 - Salario del empleado

Un empleado trabaja 48 horas en la semana a razón de $5.000 hora. El porcentaje de retención en la fuente es del 12,5% del salario bruto. Se desea saber cuál es el salario bruto, la retención en la fuente y el salario neto del trabajador.

````python
class Empleado:

    def __init__(self, horas_trabajadas, valor_hora, porcentaje_retencion):
        self.horas_trabajadas = horas_trabajadas
        self.valor_hora = valor_hora
        self.porcentaje_retencion = porcentaje_retencion

    def salario_bruto(self):
        return self.horas_trabajadas * self.valor_hora

    def retencion_fuente(self):
        return self.salario_bruto() * (self.porcentaje_retencion / 100)

    def salario_neto(self):
        return self.salario_bruto() - self.retencion_fuente()


horas = float(input("Ingrese las horas trabajadas en la semana: "))
valor_hora = float(input("Ingrese el valor de la hora: "))
porcentaje = float(input("Ingrese el porcentaje de retención en la fuente: "))

empleado = Empleado(horas, valor_hora, porcentaje)

print("Salario bruto:", empleado.salario_bruto())
print("Retención en la fuente:", empleado.retencion_fuente())
print("Salario neto:", empleado.salario_neto())
````

### Ejercicio Propuesto N° 14 - Cuadrado y cubo de un número

Elabore un algoritmo que lea un número y obtenga su cuadrado y su cubo.

````python
class Numero:

    def __init__(self, valor):
        self.valor = valor

    def cuadrado(self):
        return self.valor ** 2

    def cubo(self):
        return self.valor ** 3


valor = float(input("Ingrese un número: "))

numero = Numero(valor)

print("El cuadrado de", numero.valor, "es:", numero.cuadrado())
print("El cubo de", numero.valor, "es:", numero.cubo())
````

### Ejercicio Propuesto N° 17 - Área y perímetro de un círculo

Dado el radio de un círculo. Haga un algoritmo que obtenga el área del círculo y la longitud de la circunferencia.

````python
import math


class Circulo:

    def __init__(self, radio):
        self.radio = radio

    def area(self):
        return math.pi * self.radio ** 2

    def longitud_circunferencia(self):
        return 2 * math.pi * self.radio


radio = float(input("Ingrese el radio del círculo: "))

circulo = Circulo(radio)

print("Área del círculo:", circulo.area())
print("Longitud de la circunferencia:", circulo.longitud_circunferencia())
````
